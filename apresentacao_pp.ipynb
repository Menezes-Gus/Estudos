### <b>Kolmogorov–Smirnov (KS) </b>
##### Métrica utilizada usualmente em modelos de classificação binária. Mede a separação entre as distribuições, buscando o ponto onde a diferença entre as distribuições é máxima. Varia de 0 até 1, mas, realisticamente falando, algo acima de 0,3 ja é bom.
##### Exemplo, se assumirmos uma tabela com score de pagamento, e uma flag de caloteiro ou bom pagador (1 e 0 respectivamente), podemos calcular a estatística KS de seguinte forma:
##### 1- ordenamos os dados por score (que pode ter sido o resultado de um modelo, vide, o processo de avaliação)
| Cliente | Score | Real    |
| ------- | ----- | ------- |
| A       | 0.95  | 1       |
| B       | 0.90  | 1       |
| C       | 0.85  | 0       |
| D       | 0.80  | 1       |
| E       | 0.70  | 0       |
| F       | 0.60  | 0       |
| G       | 0.50  | 1       |
| H       | 0.40  | 0       |
| I       | 0.30  | 0       |
| J       | 0.10  | 0       |
##### 2- percorremos a lista ordenada, atribuindo um percentual de quantos % da classe 1 já apareceu na lista até aquela posição, e quantos % da classe 0 apareceu até aquela posição.
| Ponto    | % classe 1 acumulada | % classe 0 acumulada |
| -------- | ---------------- | ---------------- |
| A (0.95) | 25%              | 0%               |
| B (0.90) | 50%              | 0%               |
| C (0.85) | 50%              | 16.7%            |
| D (0.80) | 75%              | 16.7%            |
| E (0.70) | 75%              | 33.3%            |
| F (0.60) | 75%              | 50%              |
| G (0.50) | 100%             | 50%              |

##### 3- Subtraimos o percentual de uma classe da outra, em módulo.
| Ponto    | % classe 1 acumulada | % classe 0 acumulada | Diferença |
| -------- | ---------------- | ---------------- | --------- |
| A (0.95) | 25%              | 0%               | 25%       |
| B (0.90) | 50%              | 0%               | 50%       |
| C (0.85) | 50%              | 16.7%            | 33.3%     |
| D (0.80) | 75%              | 16.7%            | 58.3%     |
| E (0.70) | 75%              | 33.3%            | 41.7%     |
| F (0.60) | 75%              | 50%              | 25%       |
| G (0.50) | 100%             | 50%              | 50%       |

##### 4- A maior diferença indica quando a separação entre as 2 classes é maior, ou seja, indica o maior threshold. No nosso exemplo, equivaleria a qual o limiar de score que gera a separação máxima dos dados analisados. KS = 58.3%,threshold ideal seria um score de 80%.




### <b> AUC (Area Under the Curve ou Area Sob a Curva) </b>
##### Outro método para avaliação de modelos de classificação binária. A curva da qual a area se refere é a curva ROC (Receiver Operating Characteristic). A ROC, por sua vez, é calculada traçando uma curva, onde o eixo Y é o resultado da Sensibilidade em cada threshold possivel do modelo, enquanto que no eixo X é o resultado de 1 - Especificidade em cada threshold possivel do modelo.
##### <b>Sensibilidade (ou recall)</b>, por sua vez, indica quantos positivos foram classificados como positivo dentro da "população" de positivos da amostra.
##### <b>1 - Especificidade</b>, por sua vez, indica quantos negativos foram classificados como positivo dentro da "população" de negativos.
##### A AUC pode variar de 0 até 1, onde, o valor 0.5 indica aleatoriedade total (o modelo não é capaz de separar os dados melhor do que literalmente o acaso), e valores diferentes de 0.5 indicam graus diferentes de separação. Interessante destacar que, valores abaixo de 0.5 indicam inversão de classificação. Acima de 0.7 o modelo já pode ser consideravel razoavel, e acima de 0.8 bom.
##### O valor de AUC pode ser interpretado também como a probabilidade de ocorrencia do positivo.
##### Seu calculo se da literalmente como um calculo de área.


### <b>Gini</b>
##### Também utilizado para avaliar modelos de classificação binária. É essencialmente uma função da AUC, sendo: <i>Gini = 2 x AUC - 1</i>. Seu resultado varia de 0 até 1, onde valores acima de 0.4 já são considerados bons.
##### O indice de Gini é derivado da economia, utilizado para medir desigualdade (que é o que buscamos em uma classificação binaria, um alto valor de desigualdade).
##### Em última instancia, mede a distancia entre uma curva aleatória média e a curva do modelo encontrado. A curva diagonal média indica uma probabilidade de 50%, ou seja, em qualquer ponto da curva, se selecionarmos uma amostra aleatoria, ela tem 50% de chance de ser 0 e 50% de chance de ser 1. A medida que temos uma AUC distinta da curva diagonal média (ou seja, com um indice de gini superior a 0), mexemos nessa probabilidade, indicando um ponto ideal de separação para o modelo.



### <b>RMSE (Root Mean Squared Error ou Raiz do Erro Quadrático Médio)</b>
##### Utilizada em modelos de regressão, mede, na mesma unidade dos dados, o quanto, em média, as previsões do modelo estão distantes dos valores reais. Exibe sensibilidade a outliers, tornando-a indicada quando existe desejo/necessidade de penalizar grandes erros.
##### Deriva-se da MSE (Mean Squared Error), que eleva o erro ao quadrado (afim de evitar que a soma dos erros positivos e negativos resulte em 0).


### <b>MAE (Mean Absolute Error ou Erro Médio Absoluto)</b>
##### Utilizada em modelos de regressão, mede, na mesma unidade dos dados, em média, a distancia entre valores previstos e reais, só que, de forma linear, não atribuindo pesos maiores a outliers, por exemplo. A mesma é, então, menos sensível a outliers, o que pode ser desejavel ou não dependendo do que se esta modelando.


### <b>Accuracy</b>
##### É uma métrica para modelos de classificação, no geral. Em suma, mede o quanto de acerto, no geral, o modelo teve (ou seja, classificou a classe corretamente).
##### Isso implica em limitações, que as métricas precision e recall visam detectar. É possível se obter uma acuracia grande em casos onde as classes estão desbalanceadas, o que vai erroneamente indicar que o modelo é bom, quando, algumas vezes, uma das classes pode estar sendo negligenciada.


### <b>Precision</b>
##### É uma métrica utilizada em modelos de classificação. Mede o percentual de verdadeiros positivos determinados entre todas as amostras classificadas como positivas. É bastante útil quando detectar um falso positivo é algo indesejado.


### <b>Recall</b>
##### É também uma métrica utilizada em modelos de classificação. Mede o percentual de identificações positivas dentro da "população" positiva da amostra. Ou seja, de tudo que realmente era positivo, quantos foram classificados corretamente como positivos. Excepcionalmente útil quando a classificação de um falso negativo é indesejada.


### <b>F1-Score</b>
##### É uma métrica utilizada em modelos de classificação. Essencialmente, é uma média harmonica entre precision e recall. É uma média harmonica porque isso implica em uma maior punição em casos onde uma das 2 métricas (precisio e recall) é muito baixa. Varia de 0 a 1.
##### Precision e Recall usualmente tem uma relação de trade-off, proporcionalidade inversa. Por isso, é interessante maximizar o F1-Score, afim de equilibrar ambos.

### <b>R2 ou Coeficiente de Determinação</b>
##### É uma métrica que mede o quanto da variação dos valores reais um determinado modelo de REGRESSÃO é capaz de explicar. Pode variar de -1 até 1, onde um valor negativo indica que o modelo é pior em explicar a variação dos dados do que uma média.
##### É determinado como 1 - a divisão entre SSR (Soma dos quadrados dos resíduos), que é o erro do modelo, e SST (soma total dos quadrados), que é o erro "baseline", erro calculado entre o valor real e a média.


### <b>Validações:</b>
#### <b>Holdout</b> 
##### Validação que consiste na divisão do dataset em 2 ou 3 grupos, sendo eles: treino, teste (e, se houver uma terceira divisão, validação). Tem como vantagem sua simplicidade, e é desvantajoso em datasets muito pequenos ou onde existe dependencia temporal entre os dados.
#### <b>Leave One Out</b>
##### Esse método de validação é derivado de validação cruzada (cross validation). Nesse caso, treina-se utilizando todo o dataset menos 1 entrada, e então, o teste é feito apenas encima dessa entrada. Isso é feito para todas as entradas do dataset. Depois, tira-se a média da performance individual, obtendo a performance média do modelo. Sua desvantagem é que é computacionalmente custoso, o que pode se tornar problematico em datasets muito grandes, e, é sensível a outliers.
#### <b>K-Fold Cross Validation</b>
##### Essa validação consiste em uma divisão do dataset em partes iguais, onde utiliza-se todas as partes menos 1 para treino, e a parte restante para teste (ou validação). Tem a vantagem de ser mais rápido e menos sensível que a Leave One Out, e também de ser mais estável e preciso que a validação holdout, funcionando como um meio termo entre as duas.


#### <b>Out of Sample</b>
##### Essencialmente, consiste em testar o modelo com dados fora da amostra utilizada para treino. Assim como todas as outras formas de validação apresentadas até então, tem como foco evitar overfitting do modelo.
#### <b>Out of Time</b>
##### Similar a validação Out of Sample, a validação Out of Time é uma versão adaptada para dados com componentes temporais, ou seja, onde a ordem dos dados importa. Outros metodos de validação podem apresentar data leakage, onde, dados do futuro podem interferir em dados do passado durante o treinamento do modelo. Dessa forma, os dados utilizados para validação/teste são dados do futuro, quando comparados a dados utilizados no treinamento, na mesma unidade de tempo.


### <b>Tratamento de Missings</b>
#### Tratamento de Missings é um assunto muito abrangente. Muito provavelmente o que é apresentado aqui não esgota-o.
#### <b>Com relação ao entendimento dos dados faltantes, pode-se dividir em 3 categorias:</b>
##### <b>MCAR (Missing Completly at Random):</b> ausencia totalmente ao acaso. É o cenario mais seguro para se imputar dados.
##### <b>MAR (Missing at Random):</b> a falta é dependente de outra variavel (por exemplo, uma base onde a coluna de renda é faltante em casos onde há baixa escolaridade).
##### <b>MNAR (Missing not at Random):</b> a falta depende da propria variável (por exemplo, quem tem uma renda muito alta prefere não informar).

##### Uma vez entendido qual cenário de dados faltantes se tem, é necessario alguns pré tratamentos antes de uma estratégia de tratamento de missings de fato.
##### Identificar o percentual de faltantes em colunas e linhas, identificar se existem padrões (que podem implicar em MAR ou MNAR), tratar códigos sentinelas (como -1, 999, 'N/A', etc), e, em casos de séries temporais, identificar janelas continuas de ausencia e sazonalidade.

#### <b>Agora, para o tratamento de fato:</b>
##### <b>Remoção de linhas:</b> válido quando o percentual de linhas afetadas é baixo e for identificada MCAR
##### <b>Remoção de colunas:</b> válido quando a variável em questão não for relevante ou crítica para análise
##### <b>Imputação Numérica:</b> quando imputa-se a mediana (robusta a outliers) ou média (quando identificada distribuição proxima a normal). Também pode-se imputar uma constante e adicionar uma coluna de flag indicando dado faltante. Tem como possível revés a subestimação da variância dos dados.
##### <b>Imputação Categórica:</b> quando imputa-se a moda, ou alguma categoria distinta das presentes no campo, indicando dado faltante (podendo ser pareada com coluna de flags também).Tem como possível revés a subestimação da variância dos dados.
##### <b>Imputação Agrupada:</b> imputação que pode ser feita ccom as tecnicas anteriores, mas, condicionada a um subgrupo.Tem como possível revés a subestimação da variância dos dados.
##### <b>KNN Imputer:</b> usa KNN para identificar um valor a ser imputado
##### <b>Regressão e Classificação:</b> literalmente prevê o valor faltante com base nas outras variaveis
##### <b>Iterative Imputer/MICE:</b> é um método iterativo que, em ultima instancia, vai criar modelos que vão estimar os valores da variavel faltante, e isso sera feita iterativamente, de forma a aproximar o valor estimado ao real. As iterações param depois de um valor determinado ou quando os valores estimados não mudarem mais. Usa usualmente regressão logística ou linear.
##### <b>MissForest:</b> similar ao MICE, só que com modelo de random forest
##### <b>Matrix Factorization:</b> assumem que os dados tem estrutura de baixa dimensionalidade, ou seja, que existem pouquíssimos fatores que relacionam as variaveis conhecidas e a variavel com missings. Então, no caso de Matrix factorization, a matriz é decomposta em duas matrizes menores, e o produto das matrizes reconstroi a matriz original com os valores preenchidos.
##### <b>EM (Expectation-Maximization):</b> alterna entre E-step (expectation) , onde chuta os parametros iniciais, calculando vetores esperados dos missings, e M-step (maximization), onde, com os valores imputados, recalcula os parâmetros do modelo que maximizam maxima verossimilhança. Repete até convergir.
##### <b>Forward Fill e Backward fill:</b> utiliza o valor posterior ou anterior (supondo um conjunto de dados ordenado, como séries temporais).
##### <b>Interpolação:</b>
##### <b>Média Sazonal:</b> mesmo principio de imputação agrupada, só que em contexto de série temporal, e assumindo o agrupamento condicionado a sazonalidade.

##### Importante ressaltar que alguns modelos, como XGBoost, LightGBM, CatBoost e afins tratam nativamente dados faltantes.
##### Também pode ser interessante utilizar multiplas tecnicas e avaliar a performance do modelo afim de escolher a melhor.
#### <b>Guia prático:</b>
- Pouco missing (<5%): mediana/moda + flag resolve na maioria.
- 5–30%: teste mediana vs kNN vs MICE/MissForest; escolha pelo CV.
- 60% na coluna: questione a utilidade; considere dropar ou usar flag apenas.
- Séries temporais: priorize forward/backfill + interpolação e valide out-of-time.
- Modelos de árvore/boosting: tente sem imputar (nativo) + flags; compare.



### <b>Tratamento de Outliers</b>
##### Antes de iniciar tratamentos de Outliers, é necessário primeiro identificar se existe a necessidade ou não. Por exemplo, em um cenário em que se estuda a relação entre anos de estudo e salario, uma amostra com 1000 anos de estudo é claramente um outlier resultante de erros, e que deve ser tratado. Mas, por exemplo, em um cenario de fraude, o outlier pode ser justamente o que se busca identificar.
#### <b>Métodos de identificação de outliers:</b>
##### <b>Z-Score:</b> valores acima de 3 desvios padrões podem indicar outlier
##### **IQR (Interquartile Range):** dados acima e abaixo dos limites superior e inferior podem indicar outlier
##### **Boxplot, histogramas, scatterplots:** visualizações que podem mostrar, visualmente, outliers
##### **Isolation Forest:** usa como base o algoritmo de random forest, onde separa os dados com base em arvores de decisão aleatórias. Dados outliers tendem a levar menos quebras para serem isolados. Dessa forma, o modelo atribui um valor de "anomalia" para os dados baseado na facilidade de isola-los. Esse valor pode ser usado então para tirar os dados anomalos ou trata-los de alguma outra forma.
##### **DBSCAN:** agrupa os dados com base em densidade, onde regiões mais densas se tornam clusteres, e regiões pouco densas se tornam outliers. Dessa forma, os outliers vão ficar fora dos clusters, tornando facil a identificação.
##### **One Class SVM:** usa a mesma lógica de SVM para criar uma fronteira que englobe a maioria dos dados, e qualquer ponto fora dela é considerado anomalo.
#### __Métodos para tratamento de outliers:__
##### __Remoção:__ para casos em que o valor não faz sentido (como idade de 1000 anos)
##### __Transformação:__ utilização de log, log-retorno, raiz quadrada, Box-Cox, Yeo-Johnson. Funciona bem em situações onde a distribuição é assimétrica
##### __Winsorization:__ tecnica de substituição de valores extremos por limites superiores ou inferiores (IQR, Z-score, etc)
##### __Flag:__ para casos onde é desejavel modelar em favor do outlier, como no caso de fraudes, criar indicativos como uma flag de anomalia pode ajudar na modelagem
##### __Imputação:__ para casos onde houve remoção, por exemplo


### Categorização de Variáveis Continuas e Discretas
##### Categorização de uma variável é essencialmente um agrupamento dessa variável. Os benefícios são simplificação (o que pode facilitar em modelos lineares, principalmente), interpretabildiade, é uma forma de lidar com outliers, ou por questão regulatória. Entretanto, perde-se informação, pode introduzir pontos arbitrarios de corte e categorização, e não agregam muito em modelos derivados de arvore.
#### Tecnicas de Categorização:
##### __Binning uniforme (equal witdth):__ divide os dados em partes iguais, pode potencialmente criar bins com muito ou poucos dados
##### __Binning por frequencia (equal frequency):__ divide os dados de forma que cada faixa tenha a mesma quantidade de dados (quantis)
##### __Cutoffs baseados em negócio:__ faixas definidas pelo negocio (por exemplo, faixas salariais/de renda utilizadas pelo governo ou FGV)
##### __Binning supervisionado:__
##### __-Decision Tree Binning:__ usa o algoritmo de decision tree pra escolher melhores splits para separar as classes
##### __-WOE(Weight of Evidence) ou IV Binning(Information Value):__ agrupa categorias que maximizam separação entre bons e maus pagadores


### PCA (Principal Component Analysis)
##### É um método de que pode ser usado para redução de dimensionalidade.
##### Transforma um conjunto de dados correlacionados em um novo conjunto de dados, menor (no caso da redução de dimensionalidade), não correlacionados, chamados de __componentes principais__
##### Esses componentes são combinações lineares das variaveis originais, são construidos de forma a reter o máximo possível de variancia dos dados, e possuem uma hierarquia de poder explicativo, onde, o primeiro componente vai explicar a maior parte da variação dos dados, o segundo vai explicar a maior parte da variação dos dados não captada pelo primeiro, e assim sucessivamente.
##### O PCA encontra direções no espaço dos dados onde a variação é maior. Para isso, ele primeiro calcula uma matriz de covariancia (ou correlação), extrai autovalores (quanta variancia cada componente explica) e autovetores (direção dos componentes principais), ordena os componentes do maior para o menor autovalor, e projeta os dados nesses novos eixos (componentes).
##### Ele ajuda também a eliminar ruido, pode ser usado em compressão de imagens, para remover correlação entre variáveis e a tornar o algoritmo mais rapido. Entretanto, os componentes principais são combinações lineares essencialmente impossíveis de se interpretar diretamente, não capturam relações não lineares entre variaveis e podem ser sensíveis a escala dos dados.


### Correlação entre dados continuos e discretos
##### Quando falamos de dados continuos, falamos de correlação. Quando falamos de dados discretos, falamos de associação.
##### __Quando ambas são continuas:__
##### Correlação de Pearson: Mede relação linear entre variaveis
##### Correlação de Spearman: baseada em ranks, não exige linearidade, apenas monotonicidade (quando uma variavel aumenta, a outra ou só aumenta, ou só diminui)
##### Tau de Kendall: τ = (Número de Pares Concordantes - Número de Pares Discordantes) / (Número Total de Pares Possíveis), essencialmente uma votação, se houverem mais pares concordantes, o valor é maior e positivo, se houverem mais pares discordantes, o valor é menor (negativo). Os pares concordantes e discordantes são essencialmente comparações entre pares de dados de ambas as listas, onde observa-se se a relação entre os pares é concordante (Xi>Xj e Yi>Yj) ou discordante (Xi>Xj e Yi<Yj), ou empate.


##### __Quando ambas são discretas:__
##### Teste Qui-Quadrado: testa se variaveis são independentes entre si

##### Cramer's V: derivado do teste Qui-Quadrado, tem como objetivo atribuir um valor de associação, uma vez que Qui-Quadrado apenas atesta se há ou não independencia entre as variaveis. Varia de 0 a 1

##### Coeficiente Phi: similar ao Cramer's V, é utilizado quando existem 2 variaveis em ambas as listas a serem comparadas (por exemplo, sim/nao, norte/sul, true/false)



##### __Quando uma é continua e outra discreta__:
##### Point-Biserial Correlation: Mede a correlação entre uma variavel binária natural e uma variável continua. Caso a variável binária não seja natural, a forma de se calcular e o nome mudam um pouco (Point-Biserial vs Correlação Biserial).
##### ANOVA: compara médias da variável contínua entre grupos da variável categórica
##### Eta ou eta squared: mede quanto da variancia da variavel continua é explicada pela categórica
