### Kolmogorov–Smirnov (KS)
##### Métrica utilizada usualmente em modelos de classificação binária. Mede a separação entre as distribuições, buscando o ponto onde a diferença entre as distribuições é máxima. Varia de 0 até 1, mas, realisticamente falando, algo acima de 0,3 ja é bom.
##### Exemplo, se assumirmos uma tabela com score de pagamento, e uma flag de caloteiro ou bom pagador (1 e 0 respectivamente), podemos calcular a estatística KS de seguinte forma:
##### 1- ordenamos os dados por score (que pode ter sido o resultado de um modelo, vide, o processo de avaliação)
| Cliente | Score | Real    |
| ------- | ----- | ------- |
| A       | 0.95  | 1       |
| B       | 0.90  | 1       |
| C       | 0.85  | 0       |
| D       | 0.80  | 1       |
| E       | 0.70  | 0       |
| F       | 0.60  | 0       |
| G       | 0.50  | 1       |
| H       | 0.40  | 0       |
| I       | 0.30  | 0       |
| J       | 0.10  | 0       |
##### 2- percorremos a lista ordenada, atribuindo um percentual de quantos % da classe 1 já apareceu na lista até aquela posição, e quantos % da classe 0 apareceu até aquela posição.
| Ponto    | % classe 1 acumulada | % classe 0 acumulada |
| -------- | ---------------- | ---------------- |
| A (0.95) | 25%              | 0%               |
| B (0.90) | 50%              | 0%               |
| C (0.85) | 50%              | 16.7%            |
| D (0.80) | 75%              | 16.7%            |
| E (0.70) | 75%              | 33.3%            |
| F (0.60) | 75%              | 50%              |
| G (0.50) | 100%             | 50%              |

##### 3- Subtraimos o percentual de uma classe da outra, em módulo.
| Ponto    | % classe 1 acumulada | % classe 0 acumulada | Diferença |
| -------- | ---------------- | ---------------- | --------- |
| A (0.95) | 25%              | 0%               | 25%       |
| B (0.90) | 50%              | 0%               | 50%       |
| C (0.85) | 50%              | 16.7%            | 33.3%     |
| D (0.80) | 75%              | 16.7%            | 58.3%     |
| E (0.70) | 75%              | 33.3%            | 41.7%     |
| F (0.60) | 75%              | 50%              | 25%       |
| G (0.50) | 100%             | 50%              | 50%       |

##### 4- A maior diferença indica quando a separação entre as 2 classes é maior, ou seja, indica o maior threshold. No nosso exemplo, equivaleria a qual o limiar de score que gera a separação máxima dos dados analisados. KS = 58.3%,threshold ideal seria um score de 80%.




### AUC (Area Under the Curve ou Area Sob a Curva)
##### Outro método para avaliação de modelos de classificação binária. A curva da qual a area se refere é a curva ROC (Receiver Operating Characteristic). A ROC, por sua vez, é calculada traçando uma curva, onde o eixo Y é o resultado da Sensibilidade em cada threshold possivel do modelo, enquanto que no eixo X é o resultado de 1 - Especificidade em cada threshold possivel do modelo.
##### Sensibilidade (ou recall), por sua vez, indica quantos positivos foram classificados como positivo dentro da "população" de positivos da amostra.
##### 1 - Especificidade, por sua vez, indica quantos negativos foram classificados como positivo dentro da "população" de negativos.
##### A AUC pode variar de 0 até 1, onde, o valor 0.5 indica aleatoriedade total (o modelo não é capaz de separar os dados melhor do que literalmente o acaso), e valores diferentes de 0.5 indicam graus diferentes de separação. Interessante destacar que, valores abaixo de 0.5 indicam inversão de classificação. Acima de 0.7 o modelo já pode ser consideravel razoavel, e acima de 0.8 bom.
##### O valor de AUC pode ser interpretado também como a probabilidade de ocorrencia do positivo.
##### Seu calculo se da literalmente como um calculo de área.


### Gini
##### Também utilizado para avaliar modelos de classificação binária. É essencialmente uma função da AUC, sendo: <i>Gini = 2 x AUC - 1</i>. Seu resultado varia de 0 até 1, onde valores acima de 0.4 já são considerados bons.
##### O indice de Gini é derivado da economia, utilizado para medir desigualdade (que é o que buscamos em uma classificação binaria, um alto valor de desigualdade).
##### Em última instancia, mede a distancia entre uma curva aleatória média e a curva do modelo encontrado. A curva diagonal média indica uma probabilidade de 50%, ou seja, em qualquer ponto da curva, se selecionarmos uma amostra aleatoria, ela tem 50% de chance de ser 0 e 50% de chance de ser 1. A medida que temos uma AUC distinta da curva diagonal média (ou seja, com um indice de gini superior a 0), mexemos nessa probabilidade, indicando um ponto ideal de separação para o modelo.



### RMSE (Root Mean Squared Error ou Raiz do Erro Quadrático Médio)
##### Utilizada em modelos de regressão, mede, na mesma unidade dos dados, o quanto, em média, as previsões do modelo estão distantes dos valores reais. Exibe sensibilidade a outliers, tornando-a indicada quando existe desejo/necessidade de penalizar grandes erros.
##### Deriva-se da MSE (Mean Squared Error), que eleva o erro ao quadrado (afim de evitar que a soma dos erros positivos e negativos resulte em 0).


### MAE (Mean Absolute Error ou Erro Médio Absoluto)
##### Utilizada em modelos de regressão, mede, na mesma unidade dos dados, em média, a distancia entre valores previstos e reais, só que, de forma linear, não atribuindo pesos maiores a outliers, por exemplo. A mesma é, então, menos sensível a outliers, o que pode ser desejavel ou não dependendo do que se esta modelando.


### Accuracy
##### É uma métrica para modelos de classificação, no geral. Em suma, mede o quanto de acerto, no geral, o modelo teve (ou seja, classificou a classe corretamente).
##### Isso implica em limitações, que as métricas precision e recall visam detectar. É possível se obter uma acuracia grande em casos onde as classes estão desbalanceadas, o que vai erroneamente indicar que o modelo é bom, quando, algumas vezes, uma das classes pode estar sendo negligenciada.


### Precision
##### É uma métrica utilizada em modelos de classificação. Mede o percentual de verdadeiros positivos determinados entre todas as amostras classificadas como positivas. É bastante útil quando detectar um falso positivo é algo indesejado.


### Recall
##### É também uma métrica utilizada em modelos de classificação. Mede o percentual de identificações positivas dentro da "população" positiva da amostra. Ou seja, de tudo que realmente era positivo, quantos foram classificados corretamente como positivos. Excepcionalmente útil quando a classificação de um falso negativo é indesejada.


### F1-Score
##### É uma métrica utilizada em modelos de classificação. Essencialmente, é uma média harmonica entre precision e recall. É uma média harmonica porque isso implica em uma maior punição em casos onde uma das 2 métricas (precisio e recall) é muito baixa. Varia de 0 a 1.
##### Precision e Recall usualmente tem uma relação de trade-off, proporcionalidade inversa. Por isso, é interessante maximizar o F1-Score, afim de equilibrar ambos.

### R2 ou Coeficiente de Determinação
##### É uma métrica que mede o quanto da variação dos valores reais um determinado modelo de REGRESSÃO é capaz de explicar. Pode variar de -1 até 1, onde um valor negativo indica que o modelo é pior em explicar a variação dos dados do que uma média.
##### É determinado como 1 - a divisão entre SSR (Soma dos quadrados dos resíduos), que é o erro do modelo, e SST (soma total dos quadrados), que é o erro "baseline", erro calculado entre o valor real e a média.
